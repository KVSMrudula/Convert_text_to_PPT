```python
# 🔧 Install required package
!pip install -q python-pptx

# 📦 Importing required libraries
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from pptx.enum.shapes import MSO_AUTO_SHAPE_TYPE
from google.colab import files
import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime

# 🎨 Function to add a gradient background to each slide
def add_gradient_background(slide, prs):
    shape = slide.shapes.add_shape(
        MSO_AUTO_SHAPE_TYPE.RECTANGLE, 0, 0, prs.slide_width, prs.slide_height
    )
    fill = shape.fill
    fill.gradient()
    fill.gradient_stops[0].color.rgb = RGBColor(25, 25, 40)
    fill.gradient_stops[1].color.rgb = RGBColor(58, 96, 115)
    shape.line.fill.background()
    slide.shapes._spTree.insert(0, shape._element)  # Puts background behind all elements

# 🔁 Breaks long lines into shorter lines (max 10 words)
def split_long_lines(lines, max_words=10):
    wrapped = []
    for line in lines:
        words = line.strip().split()
        for i in range(0, len(words), max_words):
            wrapped.append(' '.join(words[i:i+max_words]))
    return wrapped

# 🛠️ Adds a new slide with title and either bullets or paragraph content
def create_slide(slide, title, lines, layout_mode):
    add_gradient_background(slide, prs)

    # Title box
    title_box = slide.shapes.add_textbox(Inches(1), Inches(0.5), Inches(8), Inches(1))
    p = title_box.text_frame.paragraphs[0]
    title_box.text_frame.text = title
    p.font.size = Pt(34)
    p.font.color.rgb = RGBColor(255, 255, 255)

    # Content box
    content_box = slide.shapes.add_textbox(Inches(1), Inches(1.6), Inches(8), Inches(5))
    tf = content_box.text_frame
    for i, text in enumerate(lines):
        para = tf.add_paragraph() if i > 0 else tf.paragraphs[0]
        para.text = f"• {text.strip()}" if layout_mode == "bullets" else text.strip()
        para.font.size = Pt(20)
        para.font.color.rgb = RGBColor(255, 255, 255)

# 📥 Text input and layout selection widgets
text_area = widgets.Textarea(
    value='',
    placeholder='🔤 Enter slides: Title on first line, content below.\n➡️ Separate each slide with a blank line.',
    description='Your Slides:',
    layout=widgets.Layout(width='100%', height='300px')
)

mode_selector = widgets.ToggleButtons(
    options=[('✅ Bullets', 'bullets'), ('📄 Paragraphs', 'paragraphs')],
    description='Layout:',
)

generate_button = widgets.Button(description='🎯 Generate Presentation', button_style='success')
output = widgets.Output()

# 🚀 Generate the presentation
def generate_ppt(b):
    with output:
        clear_output()
        global prs
        prs = Presentation()
        layout_mode = mode_selector.value

        blocks = [b.strip() for b in text_area.value.strip().split("\n\n") if b.strip()]
        for block in blocks:
            lines = [l.strip() for l in block.split("\n") if l.strip()]
            if not lines:
                continue
            title = lines[0]
            raw_content = lines[1:] or [""]
            content = split_long_lines(raw_content, max_words=10)
            slide = prs.slides.add_slide(prs.slide_layouts[6])
            create_slide(slide, title, content, layout_mode)

        filename = f"SlideDeck_{datetime.now().strftime('%H%M%S')}.pptx"
        prs.save(filename)
        print("✅ Presentation generated. Click below to download:")
        files.download(filename)

# 🔘 Button callback
generate_button.on_click(generate_ppt)

# 🖼️ Display the full widget interface
display(widgets.HTML("<h3>🧠 Enter Slide Content & Choose Layout</h3>"))
display(text_area, mode_selector, generate_button, output)
```
